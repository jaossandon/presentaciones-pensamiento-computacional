### Pregunta 1: Familia
1. Pregunta
¿Las familias monoparentales deberían tener el mismo reconocimiento social que las familias tradicionales?
2. Explicación
Esta pregunta busca conocer la opinión de los participantes sobre el reconocimiento social de las familias monoparentales, es decir, aquellas en las que los hijos viven principalmente con un solo padre o madre. El objetivo es identificar si las personas consideran que estas familias deberían tener el mismo reconocimiento y respeto social que las familias tradicionales. Para analizar las respuestas, utilizaremos un gráfico de barras que mostrará la cantidad de participantes que eligió cada alternativa y el porcentaje que representa cada respuesta respecto del total.

```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

pio.renderers.default = "plotly_mimetype+notebook_connected"

nombre_archivo = "Encuesta.xlsx"

datos = pd.read_excel(nombre_archivo)

print("Encuesta cargada correctamente")
print("Cantidad de respuestas:", len(datos))
```

```python

import pandas as pd
import plotly.express as px

pregunta = "Las familias monoparentales deberían tener el mismo reconocimiento social que las familias tradicionales."

respuestas = pd.to_numeric(datos[pregunta], errors="coerce").dropna()

tabla = respuestas.value_counts().reindex([1, 2, 3, 4, 5], fill_value=0)
tabla = tabla.rename_axis("Respuesta").reset_index(name="Cantidad")

alternativas = {
    1: "Totalmente en desacuerdo",
    2: "En desacuerdo",
    3: "Neutral",
    4: "De acuerdo",
    5: "Totalmente de acuerdo"
}

tabla["Alternativa"] = tabla["Respuesta"].map(alternativas)
tabla["Porcentaje"] = (
    tabla["Cantidad"] / len(respuestas) * 100
).round(1)

display(tabla)

fig = px.bar(
    tabla,
    x="Alternativa",
    y="Porcentaje",
    text="Porcentaje",
    title="Reconocimiento social de las familias monoparentales",
    labels={
        "Alternativa": "Alternativa seleccionada",
        "Porcentaje": "Porcentaje de respuestas"
    }
)

fig.update_traces(
    texttemplate="%{text}%",
    textposition="outside"
)

fig.update_layout(yaxis_range=[0, 115])
fig.show()
```

4. Conclusión del código
El código permite organizar las respuestas de la encuesta, contar cuántas personas seleccionaron cada alternativa y calcular los porcentajes correspondientes. El gráfico de barras facilita la comparación entre las distintas opiniones y permite identificar cuál fue la respuesta más frecuente. De esta manera, podremos conocer la tendencia general de los participantes respecto al reconocimiento social de las familias monoparentales y determinar si predominan las opiniones favorables, desfavorables o neutrales.




### Pregunta 2: Familia
1. Pregunta
¿El rol de proveedor económico del hogar ya no debería estar asociado a un género en particular?
2. Explicación
Esta pregunta busca conocer la opinión de los participantes sobre la relación entre el género y la responsabilidad económica dentro del hogar. Tradicionalmente, esta responsabilidad se ha asociado principalmente al hombre, pero actualmente existen diferentes formas de organización familiar en las que hombres y mujeres pueden contribuir económicamente. El análisis permitirá identificar si los participantes están de acuerdo con que esta responsabilidad no dependa de un género específico. Para ello, utilizaremos un gráfico de barras que mostrará la distribución de las respuestas y los porcentajes correspondientes a cada alternativa.

```python

pregunta = "El rol de proveedor económico del hogar ya no debería estar asociado a un género en particular."

respuestas = pd.to_numeric(datos[pregunta], errors="coerce").dropna()

tabla = respuestas.value_counts().reindex([1, 2, 3, 4, 5], fill_value=0)
tabla = tabla.rename_axis("Respuesta").reset_index(name="Cantidad")

tabla["Alternativa"] = tabla["Respuesta"].map(alternativas)
tabla["Porcentaje"] = (
    tabla["Cantidad"] / len(respuestas) * 100
).round(1)

display(tabla)

fig = px.bar(
    tabla,
    y="Alternativa",
    x="Porcentaje",
    text="Porcentaje",
    orientation="h",
    title="Opinión sobre los roles de género en la familia",
    labels={
        "Alternativa": "Alternativa seleccionada",
        "Porcentaje": "Porcentaje de respuestas"
    }
)
fig.update_traces(texttemplate="%{text}%", textposition="outside")
fig.update_layout(xaxis_range=[0, 115])
fig.show()
```

4. Conclusión del código
Este código permite analizar las opiniones de los participantes sobre la distribución de las responsabilidades económicas familiares. Mediante el conteo y el cálculo de porcentajes, podremos identificar cuántas personas están de acuerdo, en desacuerdo o mantienen una posición neutral frente a la afirmación. El gráfico facilita la comparación de las alternativas y ayuda a comprender la tendencia general de las respuestas. Con estos resultados podremos evaluar si existe una opinión mayoritaria a favor de que la responsabilidad económica del hogar no esté determinada por el género.

### Pregunta 3: Salud
1. Pregunta
¿Las listas de espera del sistema público de salud son un problema grave en Chile?
2. Explicación
Esta pregunta busca conocer la percepción de los participantes sobre la gravedad de las listas de espera en el sistema público de salud chileno. Estas listas representan el tiempo que las personas deben esperar para acceder a consultas con especialistas, exámenes o determinados procedimientos médicos. A través de las respuestas, podremos identificar cómo evalúan los encuestados esta situación y si consideran que constituye un problema importante para la población. Utilizaremos un gráfico de barras para comparar la cantidad y el porcentaje de participantes que están de acuerdo, en desacuerdo o mantienen una postura neutral.

```python

pregunta = "Las listas de espera del sistema público de salud son un problema grave en Chile."

respuestas = pd.to_numeric(datos[pregunta], errors="coerce").dropna()

tabla = respuestas.value_counts().reindex([1, 2, 3, 4, 5], fill_value=0)
tabla = tabla.rename_axis("Respuesta").reset_index(name="Cantidad")

tabla["Alternativa"] = tabla["Respuesta"].map(alternativas)
tabla["Porcentaje"] = (
    tabla["Cantidad"] / len(respuestas) * 100
).round(1)

display(tabla)

fig = px.pie(
    tabla,
    names="Alternativa",
    values="Cantidad",
    title="Opinión sobre las listas de espera en la salud pública",
    hole=0
)
fig.update_traces(textinfo="label+percent", hovertemplate="%{label}<br>Respuestas: %{value}<br>%{percent}")
fig.show()
```

4. Conclusión del código
El código permite visualizar y comparar las opiniones de los participantes sobre las listas de espera del sistema público de salud en Chile. A partir de la tabla y del gráfico, podremos identificar qué alternativa obtuvo la mayor cantidad de respuestas y qué porcentaje representa. Si predominan las respuestas de acuerdo o totalmente de acuerdo, esto indicará que la mayoría de los encuestados considera que las listas de espera son un problema grave. En cambio, si predominan las respuestas de desacuerdo, significará que existe una percepción diferente entre los participantes. La conclusión definitiva dependerá de los resultados reales obtenidos.

### Pregunta 4: Educación
1. Pregunta
¿La educación sexual integral debería ser obligatoria en todos los colegios?
2. Explicación
Esta pregunta busca conocer la opinión de los participantes sobre la incorporación obligatoria de la educación sexual integral en los establecimientos educacionales. Este tipo de educación aborda contenidos relacionados con el conocimiento del cuerpo, el respeto, las relaciones saludables, la prevención y la toma de decisiones responsables, de acuerdo con la edad de los estudiantes. El objetivo del análisis es identificar si los participantes consideran que estos contenidos deberían formar parte de la educación de todos los estudiantes. Para lograrlo, utilizaremos un gráfico de barras que mostrará la cantidad de respuestas y los porcentajes asociados a cada alternativa.

```python

pregunta = "La educación sexual integral debería ser obligatoria en todos los colegios."

respuestas = pd.to_numeric(datos[pregunta], errors="coerce").dropna()

tabla = respuestas.value_counts().reindex([1, 2, 3, 4, 5], fill_value=0)
tabla = tabla.rename_axis("Respuesta").reset_index(name="Cantidad")

tabla["Alternativa"] = tabla["Respuesta"].map(alternativas)
tabla["Porcentaje"] = (
    tabla["Cantidad"] / len(respuestas) * 100
).round(1)

display(tabla)

fig = px.line(
    tabla,
    x="Alternativa",
    y="Porcentaje",
    markers=True,
    text="Porcentaje",
    title="Opinión sobre la educación sexual integral",
    labels={
        "Alternativa": "Alternativa seleccionada",
        "Porcentaje": "Porcentaje de respuestas"
    }
)
fig.update_traces(texttemplate="%{text}%", textposition="top center")
fig.update_layout(yaxis_range=[0, max(10, tabla["Porcentaje"].max() * 1.2)])
fig.show()
```

4. Conclusión del código
El código permite analizar el nivel de aceptación o rechazo de la educación sexual integral obligatoria en los colegios. La tabla muestra la cantidad de personas que eligió cada alternativa, mientras que el gráfico facilita la comparación de los porcentajes. Si las respuestas de acuerdo y totalmente de acuerdo representan la mayor proporción, podremos concluir que existe una tendencia favorable hacia la obligatoriedad de esta educación. Por el contrario, si predominan las respuestas de desacuerdo, se observará una tendencia contraria. Los resultados permitirán describir la opinión de los participantes sin asumir una conclusión antes de analizar los datos.

Esta pregunta busca conocer la percepción de los participantes sobre la calidad de la educación pública en Chile.
